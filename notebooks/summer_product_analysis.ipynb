```python
#creating a missing value summary
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

missing = missing[missing["Missing Values"] > 0]

missing.sort_values(
    "Missing Values",
    ascending=False
)
```

```python
# Now we will check duplicate rows
df.duplicated().sum()
```

```python
#to check the actual duplicate row
df[df.duplicated()]
```

```python
# Checking unique values
# now we need to understand our categorical values
# this tells us about how many diff categories exist in each text column
categorical_columns = df.select_dtypes(
    include="object"
).columns

for column in categorical_columns:
    print(f"\n{column}")
    print("Unique values:", df[column].nunique())
```
