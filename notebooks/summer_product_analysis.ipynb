### Data Cleaning Decision — Urgency Information

The `has_urgency_banner` and `urgency_text` columns contained a large number of missing values. Investigation showed that the missing values occurred together, indicating that the products did not have an urgency banner.

Therefore:
- Missing `has_urgency_banner` values were encoded as `0`.
- Missing `urgency_text` values were represented as `"No urgency message"`.
- The relationship between the two variables was verified using a cross-tabulation.
